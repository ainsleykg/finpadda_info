## 5. Loops: repeating work
### `for` loops
A `for` loop repeats a block of code a set number of times. `range(1, 13)` produces the numbers 1 to 12, so this loop runs once per month.

```python
for month in range(1, 13):
    print(f"Month {month}: Thandi saves R500")
```

### Compound interest
Each month, Thandi's balance earns interest, then she adds R500. Interest earned on interest is **compound interest**, and a loop is the natural way to model it.

The rate below is an **assumption for illustration**, not a real product's rate.

```python
monthly_saving = 500
annual_rate = 0.08                 # assumed 8% per year
monthly_rate = annual_rate / 12
balance = 0

for month in range(1, 37):         # 36 months = 3 years
    balance = balance * (1 + monthly_rate) + monthly_saving
    if month % 12 == 0:            # print at the end of each year
        print(f"End of year {month // 12}: R{balance:,.2f}")

total_deposited = monthly_saving * 36
print(f"Deposited: R{total_deposited:,}. Interest earned: R{balance - total_deposited:,.2f}")
```

`{balance:,.2f}` formats a number with a thousands separator and two decimal places, which is how money should look.

### `while` loops
A `while` loop keeps going **as long as a condition is True**. Use it when you do not know in advance how many repeats you need. How many months until Thandi reaches R10 000?

```python
goal = 10000
balance = 0
months = 0

while balance < goal:
    balance = balance * (1 + monthly_rate) + monthly_saving
    months = months + 1

print(f"Thandi reaches R{goal:,} after {months} months (R{balance:,.2f})")
```

⚠️ If the condition never becomes False, a `while` loop runs forever. If that happens, press the ■ stop button next to the cell.

## 6. Lists and dictionaries: storing collections of data
### Lists
A **list** holds several values in order, inside square brackets. Below are a week of closing prices for a share (practice numbers, not real quotes).

```python
closing_prices = [101.50, 103.20, 102.80, 105.10, 104.60]

print(closing_prices[0])     # the first item: Python counts from 0
print(closing_prices[-1])    # the last item
print(len(closing_prices))   # how many items
print(max(closing_prices), min(closing_prices))
```

```python
average_price = sum(closing_prices) / len(closing_prices)
print(f"Average closing price: R{average_price:.2f}")

closing_prices.append(106.30)   # add Monday's price
print(closing_prices)
```

Loops and lists work beautifully together:

### Dictionaries
A **dictionary** stores **key: value** pairs inside curly brackets, like a lookup table. Here is a mini portfolio: how many units Thandi holds of each investment, and the price of each (practice numbers).

Bring everything together. Build a calculator that:
1. Uses **variables** for the monthly saving, annual interest rate, number of years and a savings goal.
2. Uses a **loop** to grow the balance month by month.
3. Stores the balance at the end of each year in a **list**.
4. Uses **if/else** to report whether Thandi reaches her goal.
5. Prints a clear summary with **f-strings**.

Follow the steps in the cell below. Change the inputs to test different scenarios.

```python
# Step 1: inputs (change these to test scenarios)
monthly_saving = 500
annual_rate = 0.08
years = 5
goal = 30000

# Step 2: set up a starting balance and an empty list for yearly balances


# Step 3: loop over every month, grow the balance and add the saving
#         at the end of each year, append the balance to your list


# Step 4: print the balance for each year


# Step 5: use if/else to say whether the goal was reached
```

```python
#@title Solution (double-click to reveal)
monthly_saving = 500
annual_rate = 0.08
years = 5
goal = 30000

balance = 0
yearly_balances = []
monthly_rate = annual_rate / 12

for month in range(1, years * 12 + 1):
    balance = balance * (1 + monthly_rate) + monthly_saving
    if month % 12 == 0:
        yearly_balances.append(balance)

for year, amount in enumerate(yearly_balances, start=1):
    print(f"Year {year}: R{amount:,.2f}")

deposited = monthly_saving * years * 12
print(f"\nDeposited R{deposited:,}; interest earned R{balance - deposited:,.2f}")

if balance >= goal:
    print(f"Goal reached: Thandi has R{balance:,.2f}, beating her R{goal:,} target.")
else:
    print(f"Goal not reached: Thandi is R{goal - balance:,.2f} short of R{goal:,}.")
```

## 8. Sneak peek: what happens next Tuesday
So far we have assumed Thandi earns the same return every year. Real markets do not work like that; returns go up and down. Next week you will learn **functions**, **pandas** and **matplotlib**, and work with real historical price data.

The cell below gives you a taste of where this is heading. You do not need to understand it yet; just run it.

```python
import random
import matplotlib.pyplot as plt

random.seed(6)
years = 10
for scenario in range(20):
    balance = 0
    path = []
    for year in range(years):
        yearly_return = random.gauss(0.10, 0.15)   # assumed average 10%, swings of about 15%
        balance = balance * (1 + yearly_return) + 500 * 12
        path.append(balance)
    plt.plot(range(1, years + 1), path, alpha=0.6)

plt.title("20 possible futures for Thandi's R500 a month")
plt.xlabel("Year")
plt.ylabel("Balance (R)")
plt.show()
```

Each line is one possible future. Running thousands of these random scenarios to understand the range of outcomes is called a **Monte Carlo simulation**, and it is exactly what you will build at the hackathon.
